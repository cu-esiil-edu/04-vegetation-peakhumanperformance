```python
import json
from glob import glob

import earthpy
import geopandas as gpd
import hvplot.pandas
import hvplot.xarray
import pandas as pd
import rioxarray as rxr
import xarray as xr
import osmnx as ox
```

```python
place_name = "Colorado State University, Fort Collins, Colorado, USA"

# fetch place boundary (returns a GeoDataFrame)
gdf = ox.geocode_to_gdf(place_name, by_osmid=False)

if gdf.empty:
    raise ValueError(f"No boundary found for: {place_name}")

# merge multipart into single geometry (optional)
merged_geom = gdf.unary_union

# create final GeoDataFrame named gdf_boundaryCSU (with same CRS as fetched data)
gdf_boundaryCSU = gpd.GeoDataFrame([{"name": place_name, "geometry": merged_geom}], crs=gdf.crs)

# inspect
print(gdf_boundaryCSU)

```

```python
gdf_boundaryCSU.hvplot(
    geo=True, tiles='EsriImagery', 
    frame_width=500,
    legend=False, fill_color=None, edge_color='white',
    # This parameter makes all the column values in the dataset visible.
    hover_cols='all')
```
